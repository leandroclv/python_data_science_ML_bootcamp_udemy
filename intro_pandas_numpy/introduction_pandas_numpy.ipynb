Adding a new column

```python
#Adding a new column with a scalar value. Adding new column to dataframe
df_exams['language score'] = 70
df_exams.head()
```

```python
#Adding a new column with an array. The dataframe has 1000 rows, so we need to create an array with 1000 elements
#Create an array of 1000 elements
language_score = np.arange(0, 1000)
```

```python
#Adding a new column to dataframe with an array
df_exams['language score'] = language_score
df_exams.head()
```

```python
#Create random integer numbers between 1 and 100
int_language_score = np.random.randint(1, 100, size = 1000)
```

```python
#min value inclusive and high value exclusive
min(int_language_score)
```

```python
#high value inclusive and high value exclusive
max(int_language_score)
```

```python
#Adding a new column to dataframe with an array
df_exams['language score'] = int_language_score
df_exams.head()
```

```python
#Create random float numbers between 1 and 100
np.random.uniform(1, 100, size=100)
```

```python
#Create random numbers for our new score columns
score_1 = np.random.randint(1,100, size=1000)
score_2 = np.random.randint(1,100, size=1000)
```

```python
#Create a series using the random numbers
series_1 = pd.Series(score_1, index=np.arange(0, 1000))
series_2 = pd.Series(score_2, index=np.arange(0, 1000))
```

```python
#Using assign() to add multiple columns
df_exams = df_exams.assign(score_1= series_1,score_2 = series_2)
df_exams.head(3)
```

```python
#Using index() to add a column at a specific position
df_exams.insert(1, 'test', series_1) #(número do índice, nome da coluna, valor da coluna)
df_exams.head(3)
```

###Sort a dataframe

```python
#Sort by one column
df_exams.sort_values(by='math score')
```

```python
#Sort descending by on column
df_exams.sort_values('math score', ascending=False)
```

```python
#Sort descending by multiple columns
df_exams.sort_values(['math score', 'reading score'], ascending=False)
```

```python
#Sort descending by multiple columns and update dataframe
#df_exams.sort_values(['math score', 'reading score'],ascending=False, inplace=True)
```

```python
#sort descending with a key function
#You can sort the column as it is now, but we'll use the 'key' argument to ignore case of text
df_exams.sort_values('race/ethnicity',ascending=True, key=lambda col:col.str.lower())
```

###set_index() and sort_index() methods

```python
import random
```

```python
#Creating non repetitive values for the index
new_index = np.arange(0, 1000)
```

```python
#shuffling the indexes
random.shuffle(new_index)
```

```python
#create a new column with new index
df_exams['new_index'] = new_index
df_exams
```

```python
#set new_index column as index
df_exams.set_index('new_index', inplace=True)
df_exams
```

```python
#sort dataframe by new index
df_exams.sort_index()
```

```python
#sort descending by index
df_exams.sort_index(ascending=False, inplace=True)
df_exams
```
