###Filter dataframe based on a condition

```python
import pandas as pd
import numpy as np
```

```python
df_laptops = pd.read_csv('laptop_price.csv', encoding='latin-1')
df_laptops.head(3)
```

```python
#Find wich rows have 'Apple' in the 'Company' column
df_laptops['Company'] == "Apple"
```

```python
#Filter dataframe based on condition
df_laptops[df_laptops['Company'] == "Apple"].head(3)
```

```python
#verify wich companies are in the 'company' column with value_counts()
df_laptops[df_laptops['Company'] == "Apple"].value_counts('Company')
```

###Create a column based on 1 conditions: np.where()

```python
#Create an array based on price tiers

np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Cheap') #(true = Expensive, false = Cheap)
```

```python
#set it to a new column
df_laptops['Price_tier'] = np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Cheap')
df_laptops.head()
```

```python
#count values in price_tier column
df_laptops['Price_tier'].value_counts()
```

###Exercise

```python
#create an array based on condition (screen size > 15)
df_laptops['Screen size'] = np.where(df_laptops['Inches'] > 15, 'Big', 'Small')
df_laptops.head()
```

```python
#count values in screen_size column
df_laptops['Screen size'].value_counts()
```

###Filter dataframe based on a multiple conditions

```python
#Find Apple laptops that cost more than 2000 euros
df_laptops[(df_laptops['Company']  == 'Apple') & (df_laptops['Price_euros'] > 2000)]
```

```python
#Find Apple laptops or  cost more than 2000 euros
df_laptops[(df_laptops['Company']  == 'Apple') | (df_laptops['Price_euros'] > 2000)]
```

```python
#Find laptops from Apple or Dell that cost over 2000 euros
df_laptops[((df_laptops['Company'] == 'Apple') | (df_laptops['Company'] == 'Dell')) & (df_laptops['Price_euros'] > 2000)].head()
```

###Creating a conditional column from more than 2 choices: np.select()

```python
#Create an array based on multiple price tiers (+2 choices)
```

```python
#Create 'conditions' and 'values' variables
conditions = [
    df_laptops['Price_euros']>3000,
    (df_laptops['Price_euros']>2000) & (df_laptops['Price_euros'] <= 3000),
    (df_laptops['Price_euros']>800) & (df_laptops['Price_euros'] <= 2000),
    df_laptops['Price_euros']<=800,
]

values = ['Too Expensive', 'Expensive', 'Affordable', 'Cheap']
```

```python
#set it to a new column
df_laptops['Price_tier'] = np.select(conditions, values, default='Unknown')
df_laptops.head()
```

###duplicated()

```python
#Find duplicates in 1 column/series
df_laptops.duplicated('laptop_ID')
```

```python
#Show elements in dataframe with duplicates in column 'laptop_ID'
df_laptops[df_laptops.duplicated('laptop_ID')]
```

```python
#Duplicates in 2 ou more columns
df_laptops.duplicated(['Product', 'TypeName', 'Inches'])
```

```python
#Show all the values duplicated (except for the first -> keep='first')
df_laptops[df_laptops.duplicated(['Product', 'TypeName', 'Inches'])]
```

```python
duplicated = df_laptops.duplicated(['Product', 'TypeName', 'Inches'])
```

```python
df_laptops[duplicated].sort_values(['Product', 'TypeName'])
```

```python
#sort dataframe asending by company and price. Cheapest first and more expensive last
df_laptops.sort_values(['Company', 'Price_euros'])
```

```python
#check all value categories in 'Company' column
df_laptops.value_counts('Company')
```

```python
#Duplicated values in 'Company' columns (default -> keep='first)
duplicated_first = df_laptops.duplicated('Company', keep='first')
```

```python
#Show dataframe with duplicated values in 'Company' column
df_laptops[duplicated_first]
```

```python
#Keep='first'(cheapest laptop per company)
#Show dataframe with non duplicated values in 'Company' column
df_laptops[~duplicated_first][['Company', 'Price_euros']]
```

```python
df_laptops[~duplicated_first].value_counts('Company')
```

```python
duplicate_last = df_laptops.duplicated('Company', keep='last')
```

```python
#Keep='last' (most expensive laptop per company)
df_laptops[duplicate_last]
```

###drop_duplicates()
